#####Diwali Sales Analysis

# Task
Analyze the Diwali sales data from "Diwali Sales Data.csv" to identify key trends and insights related to gender, age group, state, occupation, and product category.

## Load the data

### Subtask:
Load the dataset from the "Diwali Sales Data.csv" file into a pandas DataFrame.


**Reasoning**:
Load the data into a pandas DataFrame and display the first few rows and information.



```python
import pandas as pd
df = pd.read_csv('Diwali Sales Data.csv', encoding='unicode_escape')
display(df.head())
display(df.info())
```

## Clean the data

### Subtask:
Handle any missing or incorrect data.


**Reasoning**:
Drop irrelevant columns, handle missing values, and convert the 'Amount' column to the correct data type. Then display the DataFrame information to verify the changes.



```python
df.drop(['Status', 'unnamed1'], axis=1, inplace=True)
df.dropna(inplace=True)
df['Amount'] = df['Amount'].astype(int)
display(df.info())
```

## Analyze sales by gender

### Subtask:
Calculate and visualize sales by gender.


**Reasoning**:
Calculate sales by gender, print the result, and create a bar plot to visualize the sales by gender with appropriate title and labels.



```python
import matplotlib.pyplot as plt
sales_by_gender = df.groupby('Gender')['Amount'].sum()
print(sales_by_gender)
plt.figure(figsize=(6, 4))
sales_by_gender.plot(kind='bar')
plt.title('Total Sales by Gender')
plt.xlabel('Gender')
plt.ylabel('Total Sales')
plt.show()
```

## Analyze sales by age group

### Subtask:
Calculate and visualize sales by age group.


**Reasoning**:
Calculate the sum of 'Amount' for each 'Age Group', print the result, and create a bar plot to visualize the total sales by age group with appropriate labels and title.



```python
sales_by_age_group = df.groupby('Age Group')['Amount'].sum()
print(sales_by_age_group)

plt.figure(figsize=(10, 6))
sales_by_age_group.plot(kind='bar')
plt.title('Total Sales by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Total Sales')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

## Analyze sales by state

### Subtask:
Calculate and visualize sales by state.


**Reasoning**:
Group the data by state, calculate the sum of sales, sort the results, print the sorted sales by state, and then create a bar plot to visualize the sales by state with appropriate labels and title.



```python
sales_by_state = df.groupby('State')['Amount'].sum()
sales_by_state = sales_by_state.sort_values(ascending=False)
print(sales_by_state)

plt.figure(figsize=(12, 6))
sales_by_state.plot(kind='bar')
plt.title('Total Sales by State')
plt.xlabel('State')
plt.ylabel('Total Sales')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

## Analyze sales by occupation

### Subtask:
Calculate and visualize sales by occupation.


**Reasoning**:
Group the DataFrame by 'Occupation', calculate the sum of 'Amount', print the result, and create a bar plot to visualize the sales by occupation with appropriate labels and title.



```python
sales_by_occupation = df.groupby('Occupation')['Amount'].sum()
print(sales_by_occupation)

plt.figure(figsize=(12, 6))
sales_by_occupation.plot(kind='bar')
plt.title('Total Sales by Occupation')
plt.xlabel('Occupation')
plt.ylabel('Total Sales')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

## Analyze sales by product category

### Subtask:
Calculate and visualize sales by product category.


**Reasoning**:
Calculate the total sales for each product category and visualize the results using a bar plot with appropriate labels and title.



```python
sales_by_category = df.groupby('Product_Category')['Amount'].sum()
print(sales_by_category)

plt.figure(figsize=(12, 6))
sales_by_category.plot(kind='bar')
plt.title('Total Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

## Summary:

### Data Analysis Key Findings

*   Female customers accounted for significantly more sales (\$74,335,853) compared to male customers (\$31,913,276).
*   The sales analysis by age group shows the distribution of sales across different age brackets.
*   Uttar Pradesh, Maharashtra, and Karnataka were the top three states with the highest total sales.
*   The sales analysis by occupation shows the total sales amount for each of the 15 occupations present in the dataset.
*   The sales analysis by product category shows the total sales amount for each product category, indicating the sales performance of different product types.

### Insights or Next Steps

*   Investigate which age groups and states contribute most to sales to inform targeted marketing campaigns or inventory management in those regions.
