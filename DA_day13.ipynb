```python
data1 ={'ayush':np.arange(1,11),
        'sci_marks':np.random.randint(1,100,10),
        'math_marks':np.random.randint(1,100,10),
        'hindi_marks':np.random.randint(1,100,10)}
data2 ={'ayush':np.arange(5,15),
        'swiming_marks':np.random.randint(1,100,10),
        'sport_marks':np.random.randint(1,100,10),
         'music_marks':np.random.randint(1,100,10)}
        
```

```python
subject_score= pd.DataFrame(data1)
extra_score=pd.DataFrame(data2)
```

```python
subject_score
```

```python
extra_score
```

```python
extra_score.merge(subject_score,how='left')
```

```python
extra_score.merge(subject_score,how='right').fillna(0)
```

```python
extra_score.merge(subject_score,how='outer',indicator = True)
```

```python
extra_score.merge(subject_score,how='inner',indicator = True)
```

```python
extra_score.merge(subject_score,how='cross',indicator = True)
```
